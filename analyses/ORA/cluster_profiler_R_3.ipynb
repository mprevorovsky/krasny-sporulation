```r
ref <- read_delim(
    'GCF_000009045.1_ASM904v1_genomic.gff', delim='\t', comment = '#',
    col_names=c('id', 'source', 'type', 'start', 'stop', 'u1', 'strand', 'u2', 'description')
)
head(ref)
```

```r
ref$geneid <- stringr::str_extract(ref$description, '(?<=GeneID:)[0-9]+(?=;)')
ref$locustag <- stringr::str_extract(ref$description, '(?<=locus_tag=)[A-Za-z0-9_]+(?=;)')
refUniqueGL <- dplyr::distinct(ref, locustag, .keep_all=TRUE) %>% dplyr::select(geneid, locustag)
lt2gid <- split(refUniqueGL$geneid, refUniqueGL$locustag)

flt2gid <- Vectorize(function(x) {
    lt2gid[[x]]
})
```

## MAP GO to significant genes
- significant genes are those that have determined padj, have pval < 0.05 and abs(LFC) > 2.

```r
d <- '../DESeq_all'
files <- c(
    'DESeq2results_delta-EXP_vs_WT-EXP.csv',
    'DESeq2results_delta_omega-EXP_vs_WT-EXP.csv',
    'DESeq2results_delta_omega-STAT_vs_WT-STAT.csv',
    'DESeq2results_delta_omega-EXP_vs_delta-EXP.csv',
    'DESeq2results_delta_omega-STAT_vs_delta-STAT.csv',
    'DESeq2results_delta-STAT_vs_WT-STAT.csv'
)
```

```r
sig_gene_go <- list()
for (f in files) {
    df <- read_csv(paste(d, f, sep='/'), show_col_types = FALSE) %>% setNames(c("locusTag", names(.)[2:length(names(.))]))
    df$gid <- flt2gid(df$locusTag)
    # subset genes that have determined padj, have pval < 0.05 and abs(LFC) > 2.
    dfsig <- df[!is.na(df$padj) & df$padj <=0.05 & abs(df$log2FoldChange) > 2,]
    
    mapped_go <- select(org.Bsubtilis.eg.db, keys=dfsig$gid, columns=c('ENTREZID', 'GO'), keytype="ENTREZID")
    mapped_go_grouped <- mapped_go %>%
     group_by(ENTREZID) %>%
     summarise(GO = toString(GO)) %>%
     dplyr::ungroup()
    
    merged_go <- merge(dfsig, mapped_go_grouped, by.x='gid', by.y='ENTREZID', all.x=TRUE)
    
    sig_gene_go[[stringr::str_sub(f, start=15L, end=-5)]] <- merged_go
}
write.xlsx(sig_gene_go, file = 'DESeq2results_sig.xlsx', asTable = TRUE)
```

## GO over-representation analysis
- https://yulab-smu.top/biomedical-knowledge-mining-book/clusterprofiler-go.html#clusterprofiler-go-ora

```r
source('treeplot_data.r')
```

```r
to_percent <- function(x) {
    s <- sum(x)
  return(x/s)
}
```

```r
odd <- 'GO_ORA_pics/dotplots'
dir.create(odd, recursive=TRUE, showWarnings=FALSE)
odt <- 'GO_ORA_pics/treeplots'
dir.create(odt, recursive=TRUE, showWarnings=FALSE)
odp <- 'GO_ORA_pics/tree_repie'
dir.create(odt, recursive=TRUE, showWarnings=FALSE)

ll <- list()
for (f in files) {
    # remove gene ids that do not have defined adjusted pval
    # remove gene ids that have log2FoldChange NA
    print(f)
    df <- read_csv(paste(d, f, sep='/'), show_col_types = FALSE) %>% setNames(c("locusTag", names(.)[2:length(names(.))]))
    df$gid <- flt2gid(df$locusTag)
    # subset genes that have determined padj, have pval < 0.05 and abs(LFC) > 2.
    dfsig <- df[!is.na(df$padj) & df$padj <=0.05 & abs(df$log2FoldChange) > 2,]

    universe <- df[!is.na(df$padj) & !is.na(df$log2FoldChange),] %>% dplyr::pull(gid)
    
    gogroups <- c('BP', 'MF', 'CC')
    
    lll <- list()
    for (gog in gogroups) {
        ego <- enrichGO(
            gene          = dfsig$gid %>% as.character(),
            universe      = universe,
            OrgDb         = org.Bsubtilis.eg.db,
            ont           = gog,
            pAdjustMethod = "BH",
            pvalueCutoff  = 0.05,
            readable      = TRUE
        )
        
        egof <- ego %>% as.data.frame()
        
        if (nrow(egof) > 5) {
            nn <- paste(stringr::str_sub(f, start=15L, end=-5), gog, sep='_')
            
            #dotplot
            a <- dotplot(ego, showCategory=20, font.size=8) + ggtitle(paste(nn, 'top 20 GO terms'))
            ggsave(a, filename=stringr::str_interp("${odd}/${nn}.svg"))
            
            # treeplot
            # note that default clustering method is ward.D and 5 clusters (there always be 5 clusters)
            egox <- pairwise_termsim(ego, semData=godata(org.Bsubtilis.eg.db, gog))
            a <- treeplot(egox, offset.params = list(bar_tree = rel(2), hexpand=0.2))
            ggsave(a, filename=stringr::str_interp("${odt}/${nn}.svg"), width = 12)
        }

        #################################################################################
        ###                      Pie plots code                                       ###
        #################################################################################
        if (nrow(egof) > 5) {
            set_id <- stringr::str_sub(f, start=15L, end=-5)
            nn <- paste(set_id, gog, sep='_')
            egox <- pairwise_termsim(ego, semData=godata(org.Bsubtilis.eg.db, gog))
            tp_data <- treepart(egox, offset.params = list(bar_tree = rel(2), hexpand=0.2))

            a <- dplyr::tibble(go=egof$ID, Gene=I(egof$geneID %>% stringr::str_split('/')))
            a <- dplyr::inner_join(tp_data$df, a, by=c('go'='go'))
            a$cl_labels <- sapply(a$cluster_name, function(x) tp_data$cluster_label[[x]])
            gene_per_cluster <- a %>% group_by(cl_labels) %>% summarise(uniq = n_distinct(unlist(Gene)))
            
            na <- sum(sig_gene_go[[set_id]]$GO == 'NA')
            notna <- sum(sig_gene_go[[set_id]]$GO != 'NA')
            
            
            gene_freq_wip <- list()
            for (u in unique(a$cl_labels)) {
                    gene_freq_wip[[u]] <- a[a$cl_labels == u, 'Gene'] %>% unlist %>% unique()
                }
            gene_freq <- table(unlist(gene_freq_wip))
            
            calculate_unique_weighted <- function(single_list) {
              unique_elements <- unique(single_list)
              return(sum(1 / gene_freq[unique_elements]))
            }
            
            gene_per_cluster_weighted <- a %>% group_by(cl_labels) %>% summarise(uniq_weighted = calculate_unique_weighted(unlist(Gene)))
            # w/o annot
            e <- sig_gene_go[[set_id]]
            ngenes_go_not_sig <- length(setdiff(e[e$GO != 'NA', 'gene_name'], unique(unlist(a$Gene))))
            
            gene_per_cluster_w_na_wo <- gene_per_cluster_weighted %>% tibble::add_row(cl_labels='No GO', uniq_weighted=na) %>% tibble::add_row(cl_labels='GO not sig', uniq_weighted=ngenes_go_not_sig)
            
            gene_overview <- data.frame(
                'cat' = c('No GO annotation', 'num. genes without significant GO', 'num. genes with significant GO'),
                'gene_count' = c(na, ngenes_go_not_sig,length(unique(unlist(a$Gene))))
            )
            
            c1 <- ggplot(gene_overview, aes(x = "", y = gene_count, fill = factor(cat))) +
                geom_bar(width = 1, stat = "identity") + 
                geom_text(aes(label = gene_count), position = position_stack(vjust = 0.5))+
                coord_polar(theta = "y") +
                labs(fill = "Group", y = "Number of Genes", x = NULL, title = paste("DE genes w/wo GO annotation and ORA result overview")) +
                theme_void()
            
            gene_per_cluster_weighted$uniq_w_perc <- to_percent(gene_per_cluster_weighted$uniq_weighted)
            
            c2 <- ggplot(gene_per_cluster_weighted, aes(x = "", y = uniq_w_perc, fill = factor(cl_labels))) +
                geom_bar(width = 1, stat = "identity") +
                geom_text(aes(label = scales::percent(uniq_w_perc)), position = position_stack(vjust = 0.5))+
                coord_polar(theta = "y") +
                labs(fill = "Cluster", y = "Number of Unique Genes", x = NULL, title = paste("Weighted percentage of unique genes per GO Cluster significant in ORA", gog)) +
                scale_y_continuous(labels = scales::percent_format()) + 
                theme_void()
            
            p <- c1 + c2 + plot_layout(ncol = 1)
            ggsave(p, filename=stringr::str_interp("${odp}/${nn}.svg"), width = 12)
            
        }
        #################################################################################
        
        if (nrow(egof) != 0) {
            egof$ONTOLOGY <- gog
            egof$amigoLink <- paste("http://amigo.geneontology.org/amigo/term", egof$ID, sep='/')
            egof$quickgoLink <- paste("https://www.ebi.ac.uk/QuickGO/GTerm?id", egof$ID, sep='=')
            class(egof$amigoLink) <- "hyperlink"
            class(egof$quickgoLink) <- "hyperlink"
        }
        
        lll[[gog]] <- egof
        ll[[stringr::str_sub(f, start=15L, end=-5)]] <- do.call(rbind, lll)
    }
}
write.xlsx(ll, file = 'DESeq2results_sig_GO-ORA.xlsx', asTable = TRUE)
```
