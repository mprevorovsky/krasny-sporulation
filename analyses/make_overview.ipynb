```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
files = [
    'DESeq_all/DESeq2results_delta-EXP_vs_WT-EXP.csv',
    'DESeq_all/DESeq2results_delta_omega-EXP_vs_WT-EXP.csv',
    'DESeq_all/DESeq2results_delta-STAT_vs_WT-STAT.csv',
    'DESeq_all/DESeq2results_delta_omega-STAT_vs_WT-STAT.csv',
]
```

```python
data = []
for f in files:
    df2 = pd.read_csv(f).rename(columns={'Unnamed: 0':'locus'})
    name = f[len('DESeq_all/DESeq2results_'):-4]
    df2 = df2.rename(columns={'log2FoldChange':name})
    data.append(df2.loc[:,name])
```

```python
df = pd.concat(data, axis=1)
```

```python
tdf = df.transpose()
```

```python
plt.figure(figsize = (25,2))
sns.heatmap(tdf, xticklabels=False, yticklabels=True, center=0, cmap='Spectral', cbar_kws={
    "orientation": "horizontal",'location':'top', 'shrink':.5, 'label':'log2foldchange', "aspect":30
}, )
ax = plt.gca()
ax.set_xticks([0,len(df)], ['start', 'end'])
ax.set_xlabel('B. sub chr')
plt.savefig('goverview.svg')
```
